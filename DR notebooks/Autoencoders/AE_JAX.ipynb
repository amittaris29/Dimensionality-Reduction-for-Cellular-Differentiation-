# Autoencoders with JAX


```python
# JAX's syntax is (for the most part) same as NumPy's!
# There is also a SciPy API support (jax.scipy)
import jax.numpy as jnp
import numpy as np
import jax

# Special transform functions (we'll understand what these are very soon!)
from jax import grad, jit, vmap, pmap

# JAX's low level API 
# (lax is just an anagram for XLA, not completely sure how they came up with name JAX)
from jax import lax

from jax import make_jaxpr
from jax import random
from jax import device_put
import matplotlib.pyplot as plt

#For graphing 
import plotly.graph_objects as go
```

```python
def random_spherical_noise(center, radius, n, key):
    """Generate random points near a center point on the sphere with given radius."""
    subkeys = random.split(key, n)
    
    def make_point(k):
        vec = center + 0.15 * random.normal(k, shape=(3,))
        vec = radius * (vec / jnp.linalg.norm(vec))
        return vec

    noise = vmap(make_point)(subkeys)
    return noise


def generate_branch(start, direction, steps, arc_length, radius, key):
    tangent = direction / jnp.linalg.norm(direction)
    axis = jnp.cross(start, tangent)
    axis /= jnp.linalg.norm(axis)
    thetas = jnp.linspace(0, arc_length, steps)
    subkeys = random.split(key, steps)

    def rotate_and_perturb(theta, subkey):
        c, s = jnp.cos(theta), jnp.sin(theta)
        v = start
        point = (v * c +
                 jnp.cross(axis, v) * s +
                 axis * jnp.dot(axis, v) * (1 - c))
        point = point / jnp.linalg.norm(point)

        # Add tangent-space noise
        noise = random.normal(subkey, shape=(3,))
        noise -= jnp.dot(noise, point) * point
        point += 0.1 * noise
        point = point / jnp.linalg.norm(point)
        return radius * point

    return vmap(rotate_and_perturb)(thetas, subkeys)


# Parameters
radius = 5.0
root_center = radius * jnp.array([0, 0, 1])

# Root cluster

# Directions for 3 branches
branch_dirs = [
    jnp.array([1.5, 1.0, -0.5]),
    jnp.array([-1.0, 1.0, -0.5]),
    jnp.array([0.5, -1.5, -0.5])
]

main_key = random.PRNGKey(42)
key_root, *branch_keys = random.split(main_key, 5)


random_spherical_noise_jit = jit(random_spherical_noise)
root_cluster = random_spherical_noise(0.4 * root_center, radius, n=150000, key=key_root)

branches = []
terminal_clusters = []
branch_labels = []

for i, (dir_vec, key_branch) in enumerate(zip(branch_dirs, branch_keys[:3])):
    key_arc, key_term = random.split(key_branch)
    branch = generate_branch(root_center, dir_vec, 600, jnp.pi/2, radius, key_arc)
    branches.append(branch)
    branch_labels.extend([i + 1] * len(branch))

    term_center = branch[-2]
    terminal = random_spherical_noise(0.3*term_center, radius, n=500, key=key_term)
    terminal_clusters.append(terminal)
    branch_labels.extend([i + 1] * len(terminal))


# Combine all data
X = jnp.vstack([root_cluster] + branches + terminal_clusters)
labels = jnp.array([0]*len(root_cluster) + branch_labels)

# Plot
fig = go.Figure(data=go.Scatter3d(
    x=X[:, 0], y=X[:, 1], z=X[:, 2],
    mode='markers',
    marker=dict(
        size=3,
        color=labels,
        colorscale='Viridis',
        opacity=0.9
    )
))
fig.update_layout(
    title='Simulated Differentiation: Root → 3 Branches on Sphere',
    scene=dict(
        xaxis_title='X',
        yaxis_title='Y',
        zaxis_title='Z'
    ),
    margin=dict(l=0, r=0, b=0, t=40)
)
fig.show()
```

# Simple Autoencoder from Scratch

```python
# Initialize the weights and biases of the MLP based on given list of layer widths

def init_mlp_params(layer_widths):
    params = []
    for n_in, n_out in zip(layer_widths[:-1], layer_widths[1:]): #loops through pairs of consecutive layer sizes
        params.append(
            dict(weights=np.random.normal(size=(n_in, n_out)) * np.sqrt(2/n_in), #He initialization, which helps with stable training when using ReLU activation functions
                 biases=np.ones(shape=(n_out,))) #Initialize biases to one
            #for each layer create a dictionary with normally distributed weights 
        )
    return params
```

```python
#Performs a forward pass through the MLP

def forward(params, x): #x is the input data
    *hidden, last = params #Splits parameters into hidden layers(all but the last) and the last
    for layer in hidden:
        x = jax.nn.relu(jnp.dot(x, layer['weights']) + layer['biases']) #Computed dot product of x and the weights plus the bias and uses relu
    return jnp.dot(x, last['weights']) + last['biases'] #Do the same for the last layer with no activation 
```

```python
#Loss function using MSE

def loss_fn(params, x, y):
    return jnp.mean((forward(params, x) - y) ** 2)
```

```python
@jit  # notice how we do jit only at the highest level - XLA will have plenty of space to optimize
def update(params, x, y, lr):

    # Note that grads is a pytree with the same structure as params.
    # grad is one of the many JAX functions that has built-in support for pytrees!
    grads = jax.grad(loss_fn)(params, x, y)

    # Task: analyze grads and make sure it has the same structure as params

    # SGD update
    return jax.tree_multimap(
        lambda p, g: p - lr * g, params, grads  # for every leaf i.e. for every param of MLP
    )
```

```python
xs = np.random.normal(size=(128, 1))
ys = xs ** 2

# Instantiate a single input - single output, 3 layer (2 hidden layers) deep MLP
params = init_mlp_params([1, 128, 128, 1])

# Another example of how we might use tree_map - verify that shapes make sense:
jax.tree_map(lambda x: x.shape, params)

num_epochs = 5000


for _ in range(num_epochs):
    params = update(params, xs, ys, 0.0001)
```

```python
plt.scatter(xs, ys)                          # Ground truth
plt.scatter(xs, forward(params, xs))        # Model prediction
```
